## Now, we have to clean the html code removing words

- First we remove inline JavaScript/CSS
- Then we remove html comments. This has to be done before removing regular tags since comments can contain '>' characters
- Next we can remove the remaining tags

```python
# Your code
import re
```

```python
def clean_html(text):
    # rm JS
    text = re.sub(r"<script.*?>.*?</script>", " ", text, flags=re.DOTALL | re.IGNORECASE)
    # rm inline CSS
    text = re.sub(r"<style.*?>.*?</style>", " ", text, flags=re.DOTALL | re.IGNORECASE)
    # rm HTML 
    text = re.sub("<!--.*?-->", " ", text, flags=re.DOTALL)
    # rm HTML tags
    text = re.sub(r"<.*?>", " ", text)
    
    return text
```

- Remove all the special characters
    
- Remove numbers
    
- Remove all single characters
 
- Remove single characters from the start

- Substitute multiple spaces with single space

- Remove prefixed 'b'

- Convert to Lowercase

```python
# Your code
def basic_clean(text):
    text = clean_html(text)
    # rm prefixed b
    if text.startswith("b'") or text.startswith('b"'):
        text = text[2:]
    # to lowercase
    text = text.lower()

    words = text.split()
    clean_words = []

    for word in words:
        # rm special chars and nums
        word = ''.join(char for char in word if char.isalpha())

        # rm single characters and stopwords
        if len(word) > 1:
            clean_words.append(word)
    
    return clean_words
```

## Now let's work on removing stopwords
 * Remove the stopwords

```python
# Your code
english_stopwords = set(stopwords.words("english"))

def rm_sw(words):
    without_sw = [word for word in words if word not in english_stopwords]

    return without_sw
```

## Tame Your Text with Lemmatization
Break sentences into words, then use lemmatization to reduce them to their base form (e.g., "running" becomes "run"). See how this creates cleaner data for analysis!

```python
# Your code
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

def lemmatize_words(words):
    lemmatized = [lemmatizer.lemmatize(word) for word in words]

    return lemmatized
```

```python
# applying all functions
def text_process(text):
    words = basic_clean(text)
    words = rm_sw(words)
    words = lemmatize_words(words)
    return words

data_train["clean_text"] = data_train["text"].apply(text_process)
data_val["clean_text"] = data_val["text"].apply(text_process)
```

```python
def join_words(words):
    return " ".join(words)

data_train["preprocessed_text"] = data_train["clean_text"].apply(join_words)
data_val["preprocessed_text"] = data_val["clean_text"].apply(join_words)
```

## Bag Of Words
Let's get the 10 top words in ham and spam messages (**EXPLORATORY DATA ANALYSIS**)

```python
# Your code
from sklearn.feature_extraction.text import CountVectorizer
```

```python
bow_vect = CountVectorizer()

X_train_bow = bow_vect.fit_transform(data_train["preprocessed_text"])
X_val_bow = bow_vect.transform(data_val["preprocessed_text"])

print("Train BoW:", X_train_bow.shape)
print("Validation BoW:", X_val_bow.shape)
```

```python
# get 10 top words

# likely 0 = ham, 1 = spam
ham_text = data_train[data_train["label"] == 0]["preprocessed_text"]
spam_text = data_train[data_train["label"] == 1]["preprocessed_text"]
```

```python
#transform
ham_bow = bow_vect.transform(ham_text)
spam_bow = bow_vect.transform(spam_text)

# count words
ham_counts = ham_bow.sum(axis=0)
spam_counts = spam_bow.sum(axis=0)

#create dfs
ham_words = pd.DataFrame({
    "word": bow_vect.get_feature_names_out(),
    "count": ham_counts.A1
})

spam_words = pd.DataFrame({
    "word": bow_vect.get_feature_names_out(),
    "count": spam_counts.A1
})

print("Top 10 HAM words:")
display(ham_words.sort_values("count", ascending=False).head(10))

print("Top 10 SPAM words:")
display(spam_words.sort_values("count", ascending=False).head(10))
```

## Extra features

```python
# We add to the original dataframe two additional indicators (money symbols and suspicious words).
money_simbol_list = "|".join(["euro","dollar","pound","€",r"\$"])
suspicious_words = "|".join(["free","cheap","sex","money","account","bank","fund","transfer","transaction","win","deposit","password"])

data_train['money_mark'] = data_train['preprocessed_text'].str.contains(money_simbol_list)*1
data_train['suspicious_words'] = data_train['preprocessed_text'].str.contains(suspicious_words)*1
data_train['text_len'] = data_train['preprocessed_text'].apply(lambda x: len(x)) 

data_val['money_mark'] = data_val['preprocessed_text'].str.contains(money_simbol_list)*1
data_val['suspicious_words'] = data_val['preprocessed_text'].str.contains(suspicious_words)*1
data_val['text_len'] = data_val['preprocessed_text'].apply(lambda x: len(x)) 

data_train.head()
```
